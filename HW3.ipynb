## ให้สร้างฟังก์ชั่นคำนวณเวลาเป็นวินาทีของเวลาใดๆ และ print ออกมาให้สวยงาม

````
def time_to_second(time):
    t = time.split(':')
    h = int(t[0])
    m = int(t[1])
    se = int(t[2])
    total_second = (h * 3600) + (m * 60) + se
    print(f"เวลา {time} ชั่วโมง คิดเป็น {total_second:,} วินาที")
````

````
time_to_second("12:30:15")
time_to_second("12:53:15")
time_to_second("13:41:07")
time_to_second("00:59:25")
time_to_second("11:11:11")
time_to_second("16:06:09")
````

## ให้ สร้าง list ของ เวลา ```['12:30:15','13:41:07',....]``` แล้ววนลูปเรียกฟังก์ชั่นคำนวณเวลาเป็น วินาที

**commit ว่า HW3**

````
time_list = [
    "12:30:15",
    "13:41:07",
    "12:53:15",
    "00:59:25",
    "11:11:11",
    "16:06:09"]
````

````
print(f'<-------แปลงเวลา(ชั่วโมง) เป็นวินาที------->')
for i in time_list:
    time_to_second(i)
````
