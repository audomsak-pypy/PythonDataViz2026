##สร้างตารางใหม่ ที่ค่าใน ```sex``` เป็น missing ทั้งหมด

###ดึงข้อมูลมาใช้

````
import pandas as pd
import numpy as np
import os
````

````
import os
path = "/content/drive/MyDrive/DataViz_2026"
covid_file = os.path.join(path,'confirmed-cases.csv')
print(covid_file)
````

````
data = pd.read_csv(covid_file)
data.head()
````

###สร้างเป็นตารางใหม่ ที่ข้อมูลเพศเป็นค่า  NaN

````
missing_sex = data[pd.isnull(data["sex"])]
missing_sex
````

**สรุปได้ว่า:** ข้อมูลที่ missing อาจเกิดจากการเก็บข้อมูลที่ไม่ครบถ้วน การกอกข้อมูลมีข้อผิดพลาด กรอกข้อมูลตกหล่น

###ใช้คำสั่ง groupby

````
missing_sex.groupby('province_of_isolation').describe()
````
