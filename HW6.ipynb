###No.1 คำนวณ อายุเฉลี่ย ของผู้หญิง และผู้ชาย ของข้อมูลทั้งหมด

````
data_age = data_updated[['sex','age','province_of_onset']]
data_age
````

````
data_updated[data_updated['age']== -1]
````

````
famale = data_age[data_age['sex']=='หญิง']
famale
````

````
male = data_age[data_age['sex']=='ชาย']
male
````

````
famale['age'].head()
````

dtype: object

EX:
0     61.0
1     74.0
2     73.0
...  ข้อมูลยังเป็น String แปลงให้เป็น float (ex:61.0, 74.0, ...)  ก่อนการคำนวน

````
famale['age'] =  pd.to_numeric(famale['age'], errors='coerce') #แปลงให้เป็นตัวเลข
````

````
male['age'] = pd.to_numeric(male['age'])
````

````
famale_sum = 0
famale_N = 0
for age in famale['age']:
    if age > 0:
        famale_sum += age
        famale_N += 1
print("อายุเฉลี่ยผู้หญิง:", famale_sum / famale_N)
````

````
male_sum = 0
male_N = 0
for age in male['age']:
    if age > 0:
        male_sum += age
        male_N += 1
print("อายุเฉลี่ยผู้ชาย:",male_sum / male_N)
````

###No.2 คำนวณ อายุเฉลี่ย ของผู้หญิง และผู้ชาย ของผู้ป่วยในจังหวัดขอนแก่น

````
famale_KhonKaen = famale[data_age['province_of_onset']=='ขอนแก่น'] #เลือกมาเฉพาะคนที่อยู่ขอนแก่น
famale_KhonKaen
````

````
male_KhonKaen = male[data_age['province_of_onset']=='ขอนแก่น']
male_KhonKaen
````

````
famale_sum = 0
famale_N = 0
for age in famale_KhonKaen['age']:
    if age > 0:
        famale_sum += age
        famale_N += 1
print("อายุเฉลี่ยผู้หญิง:", famale_sum / famale_N)
````

````
male_sum = 0
male_N = 0
for age in male_KhonKaen['age']:
    if age > 0:
        male_sum += age
        male_N += 1
print("อายุเฉลี่ยผู้ชาย:",male_sum / male_N)
````

###No.3 หาจำนวนผู้ป่วยที่เป็นคน "คนต่างชาติเดินทางมาจากต่างประเทศ"

````
data_drop = data_updated[data_updated['nationality'] != 'Thailand'] #ตัดคนที่เป็นคนสัญชาติไทยออกทั้งหมด
data_drop
````

````
people_entry = data_drop[data_drop['risk'] == 'คนต่างชาติเดินทางมาจากต่างประเทศ']
people_entry
````

## Quiz ตัดตารางออกมาเฉพาะปี 2021 ```announce_date``` ในปี 2021

**Hint**
*   วนลูปหา index ของปี 2021
*   ตัดตารางมาเฉพาะ ปี 2021

````
data_updated["announce_date"] = pd.to_datetime(data_updated["announce_date"] ,format="%d/%m/%Y") #แปลงข้อมูลให้เป็น datetime
````

````
year_2021 = []
for i, row in data_updated.iterrows(): # ตัวแปรมารอรับสองค่า
  if row["announce_date"].year == 2021:
        year_2021.append(row)
data_2021 = pd.DataFrame(year_2021)
data_2021.head()
````
