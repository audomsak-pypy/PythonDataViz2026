เขียน function คูณ matrix

แล้ว test กับ matrix ขนาด
(2,3) * (3,2),
(4,4) * (4,1),
(2,2) * (2,2)

```python
import numpy as np
```

```python
def matric_multi(A,B):
    m = len(A)          # จำนวนแถว A
    a = len(A[0])     # จำนวนหลัก A
    b = len(B)        # จำนวนแถว B
    n = len(B[0])       # จำนวนหลัก B
    C = np.zeros((m,n))
    for i in range(m):
        for j in range(n):
            for sum in range(a):
                C[i,j] = C[i,j] + A[i,sum] * B[sum,j]
    return C
```

(2,3)*(3,2)

```python
A = np.array([[14, 12, 13],[18, 15, 13]]) # ตัวอย่างชุดตัวเลข
B = np.array([[17, 19],[14, 12],[16, 17]])
matric_multi(A, B)
```

(4,4)*(4,1)

```python
A = np.array([[4, 5, 7, 14],[12, 3, 14, 9],[3, 6, 11, 17],[11, 8, 5, 16]]) # ตัวอย่างชุดตัวเลข
B = np.array([[2],[4],[6],[8]])
matric_multi(A, B)
```

(2,2)*(2,2)

```python
A = np.array([[4, 16],[11, 24]]) # ตัวอย่างชุดตัวเลข
B = np.array([[7, 8],[9, 47]])
matric_multi(A, B)
```
